```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

xerr_map = {0: 0, 6: 0, 3: 0.5, 1.5: 0.5}

def gaussian(x, A, mu, sigma):
    return A * np.exp(-(x - mu)**2 / (2 * sigma**2)) 

def plot_dataset(ax, x, counts, title):
    # group by x position
    x_unique = np.unique(x[~np.isnan(x)])
    means, stds, xerrs = [], [], []
    for xi in x_unique:
        vals = counts[x == xi]
        vals = vals[~np.isnan(vals)]
        means.append(np.mean(vals))
        stds.append(np.std(vals))          # population std; use ddof=1 for sample
        xerrs.append(xerr_map[abs(xi)])
    means = np.array(means)
    stds  = np.array(stds)
    xerrs = np.array(xerrs)

   # gaussian fit
    p0 = [means.max(), x_unique[np.argmax(means)], 3]
    popt, pcov = curve_fit(gaussian, x_unique, means, p0=p0)
    A, mu, sigma = popt

    fit_label = (
        r"$A\,e^{-(x-\mu)^2/2\sigma^2}$" + "\n"
        f"A = {A:.3f}\n"
        f"$\\mu$ = {mu:.3f} mm\n"
        f"$\\sigma$ = {sigma:.3f} mm"
    )

    x_fit = np.linspace(x_unique.min(), x_unique.max(), 500)
    ax.plot(x_fit, gaussian(x_fit, *popt), '-', label=fit_label)
    ax.errorbar(x_unique, means, xerr=xerrs, yerr=stds, fmt='o', capsize=4)
    ax.set_xlabel('Horozontal Probe Position (mm)', fontsize = 12)
    ax.set_ylabel('Background Subtracted Counts', fontsize = 12)
    ax.set_xlim(-8, 8)
    ax.set_ylim(-1, 10)
    ax.grid(True)
    ax.set_title(title, fontsize = 14)
    ax.legend(fontsize=10)

    print(f"--- {title} ---")
    print(f"Amplitude = {A:.3f}")
    print(f"Center    = {mu:.3f} mm")
    print(f"Sigma     = {sigma:.3f} mm  (FWHM = {2.355*sigma:.3f} mm)")
    print(f"Offset    = {offset:.3f}\n")


fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# dataset 1 = your original table (fill in v_track/t as needed)
plot_dataset(axes[0], x_160_up,             counts,          r"v$_{track}$ = 160mm/s, t = 77 ms")
plot_dataset(axes[1], x_110_up,      counts_110_up,   r"v$_{track}$ = 110mm/s (up), t = 132 ms")
plot_dataset(axes[2], x_110_down,    counts_110_down, r"v$_{track}$ = 110mm/s (down), t = 370 ms")

plt.tight_layout()
plt.show()
```
