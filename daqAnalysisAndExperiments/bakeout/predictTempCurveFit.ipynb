```python
#MOT core temp 2025 DEC 10 over night

import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

# Given data points
time_data = np.array([3.6333, 3.7167, 3.7833, 3.85, 3.9667, 4.0667, 4.1833, 4.3333, 4.4833, 4.65, 4.825])
temperature_data = np.array([116.1, 116.55, 117, 117.4, 118, 118.5, 119, 119.5, 120, 120.5, 121])

# Define the exponential function
def exponential_func(x, a, b, c):
    return a * np.exp(b * x) + c
# Initial guess for the parameters
initial_guess = [0, -.1, 100]

# Perform curve fit with bounds
#bounds = ([-100, -100, -500], [200, 100, 500])  # bounds for a, b, c
params, _ = curve_fit(exponential_func, time_data, temperature_data, p0=initial_guess)#, bounds=bounds)

# Generate fitted data
fitted_time = np.linspace(time_data.min(), time_data.max()+7, 100)
fitted_temperature = exponential_func(fitted_time, *params)

# Plot data points and fitted curve
plt.plot(time_data, temperature_data, 'o', label='Data Points')
plt.plot(fitted_time, fitted_temperature, label='Exponential Fit')
plt.xlabel('Time [Hours]')
plt.ylabel('Temperature [C]')
plt.title('Time vs Temperature with Exponential Curve Fit')
plt.legend()
plt.grid()

# Add a textbox with the curve fit function and params
textbox_text = f"T(t) = a * exp(b * t) + c\na = {params[0]:.4f}\nb = {params[1]:.4f}\nc = {params[2]:.4f}"
plt.text(0.6, 0.3, textbox_text, transform=plt.gca().transAxes, bbox=dict(facecolor='white', alpha=0.7), fontsize=12)

plt.show()


# Print the fitted parameters
print("Fitted Parameters:")
print(f"a = {params[0]:.4f}")
print(f"b = {params[1]:.4f}")
print(f"c = {params[2]:.4f}")
```
